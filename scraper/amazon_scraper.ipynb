#### Import Modules

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
import time
import random
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

#### Set path for Webdriver

```python
# open the browser
browser = webdriver.Chrome()
```

```python
# load the page
browser.get("https://amazon.in")
```

```python
# setting up user input
search_query = input("Enter the product you want to search for: ")

try:
    # Wait for element to be present before interacting
    input_search = WebDriverWait(browser, 10).until(
        EC.presence_of_element_located((By.ID, "twotabsearchtextbox"))
    )
    input_search.send_keys(search_query)
    
    search_button = browser.find_element(By.XPATH, "(//input[@type='submit'])[1]")
    #search_buttot.clear()
    search_button.click()
    
    time.sleep(random.randint(2, 4))
    
except Exception as e:
    print(f"Error during search: {e}")
    browser.quit()
```
