#### Import Modules

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
import time
import random
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

#### Set path for Webdriver

```python
# open the browser
browser = webdriver.Chrome()
```

```python
# load the page
browser.get("https://amazon.in")
```

```python
# locating the search bar
input_search = browser.find_element(By.ID, "twotabsearchtextbox")
search_button = browser.find_element(By.XPATH, "(//input[@type='submit'])[1]")
```

```python
# setting up user input
search_query = input("Enter the product you want to search for: ")
```

```python
# sending user input to search bar
input_search.clear()
input_search.send_keys(search_query)
time.sleep(random.uniform(1, 3))
search_button.click()
```

```python
products_name_details = browser.find_elements(By.XPATH,"//a[contains(@class,'s-line-clamp-2')]")
rating = browser.find_elements(By.XPATH,"//span[contains(@class,'a-size-small a-color-base')]")
#print(f"{products_name_details.text} & {rating.text}")
```

```python
product_data = []

# zip() pairs each product with its corresponding rating
for name_element, rating_element in zip(products_name_details, rating):
    # Extract the text from each element
    name = name_element.text
    rate = rating_element.text
    
    # Print them out to see the progress
    print(f"{name} & {rate}")
    
    # Save them into your list as a dictionary (optional but highly recommended)
    product_data.append({
        "product_name": name,
        "rating": rate
    })
```
